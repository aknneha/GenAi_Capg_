```python
from nltk.stem import PorterStemmer
stemming=PorterStemmer()
for word in ["python","pythoner","pythoning","pythoned","pythonly"]:
    print(word + " -> " + stemming.stem(word))
```

```python
from nltk.stem import RegexpStemmer

stemmer = RegexpStemmer('ing$|ly$|ied$|s$', min=4)

words = ['playing', 'studied', 'quickly', 'books']

for word in words:
    print(word, '->', stemmer.stem(word))
```

```python
from nltk.stem import SnowballStemmer
sb=SnowballStemmer('english')

for word in ["python","pythoner","pythoning","pythoned","pythonly","finally"]:
    print(word + " -> " + sb.stem(word))
```
